```python
#Napisać funkcje generujące chmury punktów położonych w przestrzeni 3D na:
import numpy as np
min_val = 3
max_val = 16
num_points = 100
z_wys = 11
#a. płaskiej poziomej powierzchni o ograniczonej szerokości i długości
x = np.random.uniform(min_val, max_val, num_points)
y = np.random.uniform(min_val, max_val, num_points)
z = np.full(num_points,z_wys)
point = np.column_stack((x,y,z))
print(point[:5])

#b. płaskiej pionowej powierzchni o ograniczonej szerokości i wysokości
num_points_b = 100
const_y = 4
x_b = np.random.uniform(0, 13, num_points_b)
z_b = np.random.uniform(0,5, num_points_b)
y_b = np.full(num_points_b, const_y)
points_b = np.column_stack((x_b,y_b,z_b))
print(points_b[:5])
#c. powierzchni cylindrycznej o zadanym promieniu i ograniczonej wysokości
num_points_c = 100
r = 5
z_c = np.random.uniform(0, 10, num_points_c)
theta = np.random.uniform(0,2, num_points_c) * np.pi
x_c = r*np.sin(theta)
y_c = r*np.cos(theta)
point_cyl = np.column_stack((x_c, y_c, z_c))
print(point_cyl[:5])

np.savetxt('plaska_pozioma.xyz', point, fmt='%.6f')
np.savetxt('plaska_pionowa.xyz', points_b, fmt='%.6f')
np.savetxt('cylindr.xyz', point_cyl, fmt='%.6f')
```
